```python
import pandas as pd
from pathlib import Path

# Resolve path regardless of whether you run from root or notebooks/
project_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
csv_path = project_root / "data" / "processed" / "etripamil_drugs.csv"

# Load extracted drug records
df = pd.read_csv(csv_path)

# Filter out Vericiguat / Verquvo to isolate co-administered drugs
pattern = "VERICIGUAT|VERQUVO"
concomitant_df = df[~df["DRUGNAME_CLEAN"].str.contains(pattern, regex=True)]

# Display top 15 most frequent concomitant drugs
top_co_drugs = concomitant_df["DRUGNAME_CLEAN"].value_counts().head(15)
print("Top Co-administered Drugs:")
print(top_co_drugs)
```
